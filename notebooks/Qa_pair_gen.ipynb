```python
import re
import PyPDF2
from typing import List, Optional
import unicodedata
import logging

class PDFProcessor:
    def __init__(self):
        """Initialize PDFProcessor with logging setup."""
        logging.basicConfig(level=logging.INFO)
        self.logger = logging.getLogger(__name__)

    def extract_text_from_pdf(self, pdf_path: str) -> Optional[str]:
        """
        Extract text from a PDF file with error handling.
        
        Args:
            pdf_path (str): Path to the PDF file
            
        Returns:
            Optional[str]: Extracted text or None if extraction fails
        """
        try:
            text = ""
            with open(pdf_path, 'rb') as file:
                pdf_reader = PyPDF2.PdfReader(file)
                total_pages = len(pdf_reader.pages)
                
                self.logger.info(f"Processing PDF with {total_pages} pages")
                
                for page_num, page in enumerate(pdf_reader.pages, 1):
                    try:
                        page_text = page.extract_text()
                        if page_text:
                            text += page_text + "\n"
                        self.logger.info(f"Processed page {page_num}/{total_pages}")
                    except Exception as e:
                        self.logger.warning(f"Error extracting text from page {page_num}: {str(e)}")
                        continue
                        
            if not text.strip():
                self.logger.error("No text could be extracted from the PDF")
                return None
                
            return text
            
        except Exception as e:
            self.logger.error(f"Error processing PDF file: {str(e)}")
            return None

    def clean_text(self, text: str) -> str:
        """
        Clean and normalize the extracted text.
        
        Args:
            text (str): Raw text to clean
            
        Returns:
            str: Cleaned text
        """
        if not text:
            return ""
            
        # Convert to unicode and normalize
        text = unicodedata.normalize('NFKC', text)
        
        # Remove special characters but keep important punctuation
        text = re.sub(r'[^\w\s.,!?;:()\-\'"]', ' ', text)
        
        # Fix common OCR issues
        text = re.sub(r'(\d)\s*\.\s*(\d)', r'\1.\2', text)  # Fix decimal numbers
        text = re.sub(r'(\w)\s*-\s*(\w)', r'\1-\2', text)   # Fix hyphenated words
        
        # Normalize whitespace
        text = re.sub(r'\s+', ' ', text)
        
        # Fix common spacing issues around punctuation
        text = re.sub(r'\s+([.,!?;:])', r'\1', text)
        text = re.sub(r'([.,!?;:])\s+', r'\1 ', text)
        
        # Remove multiple newlines and normalize to single newline
        text = re.sub(r'\n\s*\n', '\n', text)
        
        # Remove leading/trailing whitespace
        text = text.strip()
        
        return text

    def process_pdf(self, pdf_path: str) -> Optional[str]:
        """
        Process a PDF file: extract and clean text.
        
        Args:
            pdf_path (str): Path to the PDF file
            
        Returns:
            Optional[str]: Cleaned text or None if processing fails
        """
        # Extract text from PDF
        raw_text = self.extract_text_from_pdf(pdf_path)
        if not raw_text:
            return None
            
        # Clean the extracted text
        cleaned_text = self.clean_text(raw_text)
        
        return cleaned_text

def main():
    # Example usage
    processor = PDFProcessor()
    pdf_path = "/Users/ganeshnallagachu/Desktop/world_bank/downloads/Hindi/Farming/खत अपरल 2025.pdf"
    
    # Process the PDF
    cleaned_text = processor.process_pdf(pdf_path)
    
    if cleaned_text:
        # Save cleaned text to a file
        output_file = "cleaned_farmerbook.txt"
        with open(output_file, 'w', encoding='utf-8') as f:
            f.write(cleaned_text)
        print(f"Cleaned text saved to {output_file}")
    else:
        print("Failed to process PDF")

if __name__ == "__main__":
    main()
```

```python
import json
import os
from openai import OpenAI
from dotenv import load_dotenv
from typing import List, Dict
from pydantic import BaseModel, Field

# Load environment variables
load_dotenv()

class TaggedContent(BaseModel):
    """Model for content with tags"""
    language: str = Field(description="detect the language of the text")
    region: str = Field(description="Geographical region discussed in the text, if applicable")
    crop: List[str] = Field(description="Specific crop type mentioned in the text, if any")
    tags: List[str] = Field(description="List of relevant farming tags")

class TagGenerator:
    def __init__(self):
        api_key = os.getenv('OPENAI_API_KEY')
        if not api_key:
            raise ValueError("OPENAI_API_KEY environment variable not set")
        self.client = OpenAI(api_key=api_key)

    def generate_tags(self, text: str) -> TaggedContent:
        """Generate tags for a given text chunk using LLM"""
        
        system_message = """
        You are an AI assistant specialized in agricultural text analysis. Your task is to process farming-related text chunks and extract specific tags for the text.
        Overall Goal:
        For each provided text chunk, you need to give the following tags if possible:
        1. Identify the primary language of the text.
        2. Identify the geographical region(s) discussed or relevant to the text, if this information is present.
        3. Identify and list the specific crop type(s) mentioned in the text, if any.(wheat, rice, corn,Tomato,Potato,Sunflower, etc.)
        4. If no crop is mentioned, then just return "Generic".
        4.  Topic Tags: Extract comprehensive farming-related topics including:
           - Farming practices (organic farming, irrigation, crop rotation, etc.)
           - Technology (tractors, drones, precision agriculture, etc.)
           - Livestock (cattle, poultry, dairy, etc.)
           - Business aspects (marketing, finance, cooperatives, etc.)
        
        return the output in json format
        output should be like this:
        {
            "language": "Hindi",
            "region": "India",
            "crop": ["wheat", "rice"],
            "tags": ["farming", "agriculture"]
        }
        """

        try:
            response = self.client.beta.chat.completions.parse(
                model="gpt-4o-mini",
                messages=[
                    {"role": "system", "content": system_message},
                    {"role": "user", "content": f"Analyze this agricultural text and generate tags:\n{text}"}
                ],
                response_format=TaggedContent,
                temperature=0.7,
                max_tokens=500
            )
            
            result = response.choices[0].message.parsed
            # return TaggedContent(**result)
            return result

        except Exception as e:
            print(f"Error generating tags for text: {str(e)[:100]}... - {e}")
            # Return default structure on error
            return TaggedContent(
                language="unknown", 
                region=None, 
                crop=[],
                tags=[]
            )
def process_chunks(input_file: str, output_file: str):
    """Process text chunks and add tags"""
    
    # Initialize tag generator
    tagger = TagGenerator()
    
    # Read chunks from file
    with open(input_file, 'r', encoding='utf-8') as f:
        data = json.load(f)
    chunks = [c["text"] for c in data["chunks"]]
    # Process each chunk
    tagged_chunks = []
    total_chunks = len(chunks)
    
    for i, chunk in enumerate(chunks[1:3], 1):
        # if not chunk.strip():
        #     continue
            
        # print(f"Processing chunk {i}/{total_chunks}")
        tagged_content = tagger.generate_tags(chunk)
        tagged_chunks.append(tagged_content)
        print(f'chunk :{chunk}')
        print(f"Processed chunk {i}/{total_chunks}: {tagged_content.language}, {tagged_content.region}, {tagged_content.crop}, {tagged_content.tags}")
    
    # Save tagged content
    # with open(output_file, 'w', encoding='utf-8') as f:
    #     json.dump({
    #         "total_chunks": len(tagged_chunks),
    #         "tagged_chunks": tagged_chunks
    #     }, f, indent=2, ensure_ascii=False)
    
    # print(f"Tagged {len(tagged_chunks)} chunks saved to {output_file}")
    return tagged_chunks

# if __name__ == "__main__":
#     
```

```python
"""
Chunking the output semantically
"""
import numpy as np
import re
from sklearn.metrics.pairwise import cosine_similarity
import json
import os
from openai import OpenAI
from dotenv import load_dotenv
load_dotenv()

class SemanticChunking:
    def __init__(self):
        # Get API key from environment variable
        api_key = os.getenv('OPENAI_API_KEY')
        if not api_key:
            raise ValueError("OPENAI_API_KEY environment variable not set")
            
        # Initialize OpenAI client
        self.ai_client = OpenAI(api_key=api_key)

    def calculate_cosine_distances(self, sentences: list) -> tuple[list, list]:
        n = len(sentences)
        distances = np.zeros(n - 1)
        for i in range(n - 1):
            sim = cosine_similarity(
                [sentences[i]['combined_sentence_embedding']],
                [sentences[i + 1]['combined_sentence_embedding']]
            )[0, 0]
            distance = 1 - sim
            distances[i] = distance
            sentences[i]['distance_to_next'] = distance
        return distances.tolist(), sentences

    def chunking_dp(self, distances: list, word_length=None, limit_per_chunk=10) -> list:
        if limit_per_chunk <= 0:
            raise ValueError("sentence_limit_per_chunk must be positive")
        if word_length is None:
            word_length = np.ones(len(distances) + 1, dtype=np.int32)
        else:
            word_length = np.array(word_length, dtype=np.int32)
        n = len(word_length)
        split_idx = np.full(n, n, dtype=np.int32)
        split_cost = np.full(n, np.inf)
        split_cost[-1] = 0

        for curr_idx in reversed(range(n - 1)):
            total_words = 0
            for i in range(curr_idx + 1, n):
                total_words += word_length[i]
                if total_words > limit_per_chunk:
                    break
                cost = split_cost[i] + distances[i - 1]
                if cost < split_cost[curr_idx]:
                    split_cost[curr_idx] = cost
                    split_idx[curr_idx] = i

        splits = []
        i = 0
        while i < n:
            splits.append(i)
            i = split_idx[i]
        splits.append(n)
        return splits
    
    def combine_sentences(self, sentences, buffer_size=1):
        n = len(sentences)
        for i in range(n):
            buffer = []
            if i - buffer_size >= 0:
                buffer.append(sentences[i - buffer_size]['sentence'])
            buffer.append(sentences[i]['sentence'])
            if i + buffer_size < n:
                buffer.append(sentences[i + buffer_size]['sentence'])
            sentences[i]['combined_sentence'] = ' '.join(buffer)
        return sentences

    def get_embedding_list(self, text_list):
        """
        Get embeddings for a list of text using OpenAI's API.
        Returns a list of embeddings.
        """
        results = []
        for text in text_list:
            try:
                response = self.ai_client.embeddings.create(
                    input=text,
                    model="text-embedding-ada-002"  # Use the appropriate model
                )
                results.append(response)
            except Exception as e:
                print(f"Error getting embedding: {e}")
                continue
        return results

    def semantic_grouping(self, text, max_char: int = 2000) -> list:
        """
        Process a text string into semantic chunks
        
        Parameters:
        - text: String containing the text to be processed
        - max_char: Maximum characters per chunk
        
        Returns:
        - List of dictionaries containing chunked text
        """
        if not text or not isinstance(text, str):
            return []

        # Split text into sentences
        sentences = re.split(r'(?<=[.?!])\s+', text)
        sentence_dicts = [{'sentence': s, 'index': i} for i, s in enumerate(sentences) if s]

        if not sentence_dicts:
            return []

        # Combine sentences with surrounding context
        sentence_dicts = self.combine_sentences(sentence_dicts)
        
        # Get embeddings for combined sentences
        embeddings = self.get_embedding_list(
            [x['combined_sentence'] for x in sentence_dicts]
        )

        if not embeddings:
            return []

        # Process embeddings
        for i, embedding_response in enumerate(embeddings):
            if hasattr(embedding_response, 'data') and embedding_response.data:
                sentence_dicts[i]['combined_sentence_embedding'] = embedding_response.data[0].embedding

        # Calculate distances between embeddings
        distances, sentence_dicts = self.calculate_cosine_distances(sentence_dicts)
        
        # Calculate word lengths for each sentence
        word_length = [len(s['sentence']) for s in sentence_dicts]
        
        # Get chunk indices using dynamic programming
        chunk_indices = self.chunking_dp(distances, word_length, max(max_char, max(word_length) if word_length else max_char))
        chunk_indices = chunk_indices[1:-1]  # Remove first and last indices

        # Create chunks from sentences
        chunks = []
        start_idx = 0
        for idx in chunk_indices:
            chunk_sentences = sentence_dicts[start_idx:idx + 1]
            text_chunk = ' '.join(s['sentence'] for s in chunk_sentences)
            chunks.append({'text': text_chunk})
            start_idx = idx + 1

        # Handle remaining sentences
        if start_idx < len(sentence_dicts):
            text_chunk = ' '.join(s['sentence'] for s in sentence_dicts[start_idx:])
            chunks.append({'text': text_chunk})

        return chunks
    
def main():
    # Initialize the semantic chunking processor
    chunker = SemanticChunking()
    
    
    # Process the cleaned text
    chunks = chunker.semantic_grouping(cleaned_text)
    
    # Save chunks to a JSON file
    output_file = "cleaned_text_chunks.json"
    with open(output_file, 'w', encoding='utf-8') as f:
        json.dump({
            "total_chunks": len(chunks),
            "chunks": chunks
        }, f, indent=2, ensure_ascii=False)
    
    print(f"Processed {len(chunks)} chunks from cleaned text")
    print(f"Chunks saved to {output_file}")

if __name__ == "__main__":
    main()
```

```python
import tiktoken
import json

def analyze_tokens(input_file: str, model: str = "gpt-4o-mini") -> None:
    """
    Analyze token usage for answers, evaluation results and prompts
    
    Args:
        input_file (str): Path to JSON file containing answers and evaluation results
        model (str): Model name for token calculation
    """
    # Initialize tokenizer
    encoding = tiktoken.encoding_for_model(model)
    
    # Define system and user prompts
    system_prompt = """
                              You are a fact extractor. Your task is to check how many facts are presented in a response. 
                              The response is from a chat bot to the questions by users. The questions and responses are related to farming. 
                              The responses have generally friendly tone, addressing user by name and have relevant emojis. 
                              The responses are also structured with new line and bullet points for better readability. 
                              Most of the time, it is followed by a suggested list of questions that the user can further ask.

                              You are supposed to construct a set of factual statements from the response about farming and farming alone based on following rules:
                              - Don't generate factual statement about the user, for example "Hello Jayasankar" reason being we want to extract the facts presented about farming alone
                              - Don't generate factual statements after the followup questions, for example "example questions: - how can I control pest infestation in Potato crop?" reason being the follow up questions are meant to drive conversation and not necessary mean that someone is interested in the conversation, they are just suggestions
                              - Don't generate factual statements about the response, for example "Sorry, this seems to be out of context, can you rephrase it" reason being the inability to provide answer maybe that it is actually not present in the context and this fact is not related to farming

                              If there are no factual statements about farming, return an empty list in the facts field.
                              #### Example :
                              Input :
                                Based on the information provided in the context, here is the answer to your question, Jayasankar:

                                - The coffee varieties that are resistant to Coffee Leaf Rust (CLR) include:
                                - Ruiru 11
                                - Batian
                                - K7
                                These varieties are suitable for different coffee growing areas in Kenya and are known for their resistance to CLR.
                                Hope you found the answer helpful! Would you like to ask more about coffee varieties or any other related topic?
                                Here are a few questions that may help:
                                1. What are the ideal growing conditions for coffee plants?
                                2. How can damping-off and Brown eye spot diseases be controlled in coffee plants?
                                3. Can you provide more information on the maintenance of coffee seedlings?
                                Output:
                                ["Based on the context, there are three varieties of coffee resistant to coffee leaf rust", "Ruiru 11 is a variety resistant to coffee leaf rust", "Baitan is a variety resistant to coffee leaf rust", "K7 is a variety resistant to coffee leaf rust", "The varieties are suitable to be used in Kenya"] 

                                You are a natural language inference engine. You will be presented with a set of factual statements and context. 
                                You are supposed to analyze if each statement is factually correct given the context. 
                                You can come up with the scores of 'Yes' (1) and 'No' (0) as verdict.

                                Use the following rules:
                                - If the statement can be derived from the context, give a score of 1
                                - If there is no statement and there is no context, give a score of 1
                                - If the statement can't be derived from the context, give a score of 0
                                - If there is no context but there is a statement, give a score of 0

                                For each statement, provide:
                                - The statement text
                                - A clear reason explaining your decision
                                - A verdict of '1' (supported) or '0' (not supported)
                                """
    user_prompt_template = "Evaluate if this answer is faithful to the context: {answer}"
    
    # Calculate prompt tokens
    system_tokens = len(encoding.encode(system_prompt))
    
    # Load data
    with open(input_file, 'r', encoding='utf-8') as f:
        data = json.load(f)
    
    # Initialize counters
    answer_tokens = []
    eval_tokens = []
    prompt_tokens = []
    
    # Process each entry
    for entry in data:
        if "answer" in entry and entry["answer"]:
            # Count tokens in answer
            answer_text = entry["answer"]
            answer_token_count = len(encoding.encode(answer_text))
            answer_tokens.append(answer_token_count)
            
            # Count tokens in prompt for this answer
            user_prompt = user_prompt_template.format(answer=answer_text)
            prompt_token_count = system_tokens + len(encoding.encode(user_prompt))
            prompt_tokens.append(prompt_token_count)
        
        if "evaluation_result" in entry and entry["evaluation_result"]:
            # Count tokens in evaluation result
            eval_text = str(entry["evaluation_result"])
            eval_token_count = len(encoding.encode(eval_text))
            eval_tokens.append(eval_token_count)
    
    # Calculate averages
    avg_answer_tokens = sum(answer_tokens) / len(answer_tokens) if answer_tokens else 0
    avg_eval_tokens = sum(eval_tokens) / len(eval_tokens) if eval_tokens else 0
    avg_prompt_tokens = sum(prompt_tokens) / len(prompt_tokens) if prompt_tokens else 0
    
    # Calculate totals
    total_answer_tokens = sum(answer_tokens)
    total_eval_tokens = sum(eval_tokens)
    total_prompt_tokens = sum(prompt_tokens)
    total_tokens = total_answer_tokens + total_eval_tokens + total_prompt_tokens
    
    # Print analysis
    print("\nToken Usage Analysis:")
    print("-" * 50)
    print(f"Model used: {model}")
    print(f"Total entries analyzed: {len(data)}")
    print(f"Entries with answers: {len(answer_tokens)}")
    print(f"Entries with evaluation results: {len(eval_tokens)}")
    print("\nAverage Tokens Per Entry:")
    print(f"- Prompts: {avg_prompt_tokens:.2f}")
    print(f"- Answers: {avg_answer_tokens:.2f}")
    print(f"- Evaluations: {avg_eval_tokens:.2f}")
    print("-" * 50)
    print(f"Total prompt tokens: {total_prompt_tokens:,}")
    print(f"Total answer tokens: {total_answer_tokens:,}")
    print(f"Total evaluation tokens: {total_eval_tokens:,}")
    print(f"Total tokens: {total_tokens:,}")
    print("-" * 50)
    # print(f"Estimated cost (at $0.03 per 1K tokens): ${(total_tokens/1000 * 0.03):.2f}")

if __name__ == "__main__":
    input_file = "/Users/ganeshnallagachu/Desktop/world_bank/failthfullness_results_test.json"
    analyze_tokens(input_file)
```
