```python
import requests
from bs4 import BeautifulSoup
import os
from urllib.parse import urljoin

def download_pdf(url, language, category, base_dir='downloads'):
    # Create nested directory structure
    save_dir = os.path.join(base_dir, language, category)
    if not os.path.exists(save_dir):
        os.makedirs(save_dir)
    
    try:
        # Get the webpage content with SSL verification disabled
        response = requests.get(url, verify=False)
        soup = BeautifulSoup(response.content, 'html.parser')
        
        # Find all links that end with .pdf
        pdf_links = []
        for link in soup.find_all('a'):
            href = link.get('href')
            if href and href.lower().endswith('.pdf'):
                # Convert relative URL to absolute URL
                full_url = urljoin(url, href)
                pdf_links.append(full_url)
        
        # Download each PDF
        for pdf_url in pdf_links:
            try:
                # Get the filename from the URL
                filename = os.path.join(save_dir, pdf_url.split('/')[-1])
                
                # Download the PDF with SSL verification disabled
                pdf_response = requests.get(pdf_url, verify=False)
                
                # Save the PDF
                with open(filename, 'wb') as f:
                    f.write(pdf_response.content)
                print(f"Downloaded: {filename}")
                
            except Exception as e:
                print(f"Error downloading {pdf_url}: {str(e)}")
    
    except Exception as e:
        print(f"Error processing {url}: {str(e)}")

# Process each dictionary of links
# all_categories = {
#     "English": English,
#     "Hindi": Hindi,
#     "Traditional_Knowledge": Traditional_knowledge_english
# }

# for language, categories in all_categories.items():
#     for category, url in categories.items():
#         print(f"\nProcessing: {language}/{category}")
#         download_pdf(url, language, category)
```

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import time
import os

def scrape_journal_pdfs(journal_links, base_dir='journal_downloads'):
    try:
        for journal, links in journal_links.items():
            print(f"\nProcessing Journal: {journal}")
            # Create journal specific directory
            journal_dir = os.path.join(base_dir, journal.replace(" ", "_"))
            if not os.path.exists(journal_dir):
                os.makedirs(journal_dir)
            
            # Set up Chrome options with journal-specific download directory
            chrome_options = webdriver.ChromeOptions()
            chrome_options.add_experimental_option('prefs', {
                "download.default_directory": os.path.abspath(journal_dir),
                "download.prompt_for_download": False,
                "plugins.always_open_pdf_externally": True,
                'download.directory_upgrade': True,
                'safebrowsing.enabled': True
            })
            
            # Initialize driver for each journal with its specific download directory
            driver = webdriver.Chrome(options=chrome_options)
            wait = WebDriverWait(driver, 10)
            
            for issue_url in links:
                try:
                    print(f"\nProcessing issue: {issue_url}")
                    driver.get(issue_url)
                    time.sleep(3)  # Wait for page to load
                    
                    # Step 1: Get all article PDF links
                    article_links = driver.find_elements(By.CSS_SELECTOR, "a.obj_galley_link.pdf")
                    article_hrefs = [link.get_attribute('href') for link in article_links]
                    
                    # Step 2: Visit each article page and get download link
                    for article_href in article_hrefs:
                        try:
                            print(f"Processing article: {article_href}")
                            driver.get(article_href)
                            time.sleep(2)
                            
                            # Find and click download button
                            download_link = wait.until(
                                EC.presence_of_element_located((By.CSS_SELECTOR, "a.download"))
                            )
                            download_href = download_link.get_attribute('href')
                            
                            # Visit download URL which will trigger automatic download
                            driver.get(download_href)
                            time.sleep(3)  # Wait for download to start
                            
                            print(f"Downloaded PDF from article: {article_href}")
                            
                        except Exception as e:
                            print(f"Error processing article {article_href}: {str(e)}")
                            continue
                            
                except Exception as e:
                    print(f"Error processing issue {issue_url}: {str(e)}")
                    continue
                    
            # Close the driver for this journal
            driver.quit()
                    
    except Exception as e:
        print(f"Error in main process: {str(e)}")
        if 'driver' in locals():
            driver.quit()

# Run the scraper
scrape_journal_pdfs(journal_links)
```

```python
import requests
from bs4 import BeautifulSoup
from urllib.parse import urljoin
import os

def scrape_hindi_pdfs(links_dict):
    base_dir = '/Users/ganeshnallagachu/Desktop/world_bank/downloads/Hindi'
    
    for category, urls in links_dict.items():
        # Create category directory
        save_dir = os.path.join(base_dir, category)
        if not os.path.exists(save_dir):
            os.makedirs(save_dir)
            
        for url in urls:
            print(f"\nProcessing URL: {url}")
            try:
                # Get the webpage content
                response = requests.get(url, verify=False)
                soup = BeautifulSoup(response.content, 'html.parser')
                
                # Find all anchor tags that contain PDF links
                pdf_links = soup.find_all('a', href=lambda x: x and x.endswith('.pdf'))
                
                for link in pdf_links:
                    try:
                        # Get the PDF URL
                        pdf_url = link.get('href')
                        if pdf_url.startswith('/'):
                            pdf_url = urljoin('https://icar.org.in', pdf_url)
                            
                        # Get the filename from alt text of image if available, otherwise from URL
                        img = link.find('img')
                        if img and img.get('alt'):
                            filename = f"{img.get('alt')}.pdf"
                        else:
                            filename = pdf_url.split('/')[-1]
                            
                        # Clean filename of any invalid characters
                        filename = "".join([c for c in filename if c.isalnum() or c in (' ', '-', '_', '.')]).rstrip()
                        filepath = os.path.join(save_dir, filename)
                        
                        # Download the PDF
                        pdf_response = requests.get(pdf_url, verify=False)
                        if pdf_response.status_code == 200:
                            with open(filepath, 'wb') as f:
                                f.write(pdf_response.content)
                            print(f"Downloaded: {filename}")
                        else:
                            print(f"Failed to download: {pdf_url}")
                            
                    except Exception as e:
                        print(f"Error downloading PDF {pdf_url}: {str(e)}")
                        
            except Exception as e:
                print(f"Error processing URL {url}: {str(e)}")

# Run the scraper
scrape_hindi_pdfs(Hindi_links)
```
