```python
import pandas as pd
import json
import ast

def parse_qa_pairs(qa_json_str, chunk_id_str):
    """Parse QA pairs with robust error handling"""
    try:
        # Parse the JSON string
        qa_data = json.loads(qa_json_str)
        
        # Split chunk_id string and create a list of chunk IDs
        chunk_ids = [chunk.strip() for chunk in chunk_id_str.split(' + ')]
        
        # Create list to store parsed data
        parsed_data = []
        
        for qa_pair in qa_data['qa_pairs']:
            question = qa_pair['question']
            chunks_used = qa_pair['chunks_used']
            
            # Parse chunks_used (e.g., "3, 4" -> [3, 4])
            chunk_indices = [int(x.strip()) for x in chunks_used.split(',')]
            
            # Map indices to actual chunk IDs (subtract 1 for 0-based indexing)
            mapped_chunk_ids = [chunk_ids[i-1] for i in chunk_indices if i-1 < len(chunk_ids)]
            # remove "chunk" from the chunk_ids
            mapped_chunk_ids = [id.replace("chunk_", "") for id in mapped_chunk_ids]
            
            parsed_data.append({
                'question': question,
                'chunk_ids': mapped_chunk_ids
            })
        
        return parsed_data
        
    except json.JSONDecodeError as e:
        print(f"JSON decode error at position {e.pos}: {e.msg}")
        print(f"Error context: {qa_json_str[max(0, e.pos-50):e.pos+50]}")
        return []
    except Exception as e:
        print(f"Other error: {e}")
        return []

# Process the dataframe with better error handling
all_parsed_data = []
successful_rows = 0
failed_rows = 0

print("Processing rows...")
for idx, row in df.iterrows():
    if idx % 1000 == 0:
        print(f"Processing row {idx}/{len(df)}")
    
    qa_pairs_str = row['QA_pairs']
    chunk_id_str = row['chunk_id']
    
    parsed_data = parse_qa_pairs(qa_pairs_str, chunk_id_str)
    if parsed_data:
        all_parsed_data.extend(parsed_data)
        successful_rows += 1
    else:
        failed_rows += 1
        if failed_rows <= 5:  # Only show first 5 failures
            print(f"Failed to parse row {idx}")

print(f"\nProcessing complete!")
print(f"Successful rows: {successful_rows}")
print(f"Failed rows: {failed_rows}")
print(f"Total QA pairs extracted: {len(all_parsed_data)}")

# Create new dataframe
if all_parsed_data:
    new_df = pd.DataFrame(all_parsed_data)
    
    # Display the result
    print("\nParsed DataFrame:")
    print(new_df.head())
    print(f"\nShape: {new_df.shape}")
    
    # Save to CSV
    new_df.to_csv('parsed_qa_chunks1.csv', index=False)
    print("\nCSV file 'parsed_qa_chunks.csv' has been created!")
    
    # Show first few rows with chunk_ids as strings for CSV compatibility
    print("\nFirst few rows:")
    for idx, row in new_df.head().iterrows():
        print(f"Question: {row['question'][:50]}...")
        print(f"Chunk IDs: {row['chunk_ids']}")
        print("-" * 50)
else:
    print("No data was successfully parsed!")
```

```python
import pandas as pd
import numpy as np
from qdrant_client import QdrantClient
from qdrant_client.models import Filter, FieldCondition, MatchValue
from openai import OpenAI
import time
from tqdm import tqdm
from concurrent.futures import ThreadPoolExecutor, as_completed
import threading

def get_embedding(text, openai_client, model="text-embedding-3-small"):
    """Get embedding for text using OpenAI."""
    try:
        response = openai_client.embeddings.create(
            input=text,
            model=model
        )
        return response.data[0].embedding
    except Exception as e:
        print(f"Error getting embedding: {e}")
        return None

def retrieve_chunks_for_question(question, client, collection_name, openai_client, top_k=10):
    """Retrieve top-k chunks for a given question."""
    try:
        # Get embedding for the question
        question_embedding = get_embedding(question, openai_client)
        if question_embedding is None:
            return question, []
        
        # Search in Qdrant
        search_result = client.search(
            collection_name=collection_name,
            query_vector=question_embedding,
            limit=top_k,
            with_payload=True
        )
        
        # Extract chunk IDs from search results
        chunk_ids = []
        chunks_list = []
        for result in search_result:
            # if 'payload' in result and 'chunk_id' in result.payload:
            chunk_ids.append(result.payload['chunk_id'])
            chunks_list.append(result.payload['chunk_text'])
        
        return question, chunk_ids , chunks_list
        
    except Exception as e:
        print(f"Error retrieving chunks for question: {e}")
        return question, []

def process_question_batch(question_batch, client, collection_name, openai_client, top_k=5):
    """Process a batch of questions in parallel."""
    results = []
    
    # Create a thread-local OpenAI client for each thread
    thread_local = threading.local()
    if not hasattr(thread_local, 'openai_client'):
        thread_local.openai_client = OpenAI(api_key=openai_client.api_key)
    
    for question in question_batch:
        question, chunk_ids , chunks_list = retrieve_chunks_for_question(
            question, client, collection_name, thread_local.openai_client, top_k
        )
        results.append((question, chunk_ids , chunks_list))
        time.sleep(0.05)  # Small delay to avoid rate limiting
    
    return results

def parallel_retrieve_chunks(questions, client, collection_name, openai_client, max_workers=10, batch_size=50, top_k=5):
    """Retrieve chunks for questions using parallel processing."""
    all_results = []
    
    # Split questions into batches
    question_batches = [questions[i:i + batch_size] for i in range(0, len(questions), batch_size)]
    
    print(f"Processing {len(questions)} questions in {len(question_batches)} batches with {max_workers} workers...")
    
    with ThreadPoolExecutor(max_workers=max_workers) as executor:
        # Submit all batches
        future_to_batch = {
            executor.submit(process_question_batch, batch, client, collection_name, openai_client, top_k): batch 
            for batch in question_batches
        }
        
        # Process completed batches with progress bar
        with tqdm(total=len(question_batches), desc="Processing batches") as pbar:
            for future in as_completed(future_to_batch):
                try:
                    batch_results = future.result()
                    all_results.extend(batch_results)
                    pbar.update(1)
                    pbar.set_postfix({'Processed': len(all_results)})
                    
                except Exception as e:
                    print(f"Error in batch processing: {e}")
                    pbar.update(1)
    
    return all_results

def main():
    # Initialize clients
    print("Initializing clients...")
    qdrant_client = QdrantClient(url="http://dev.platform.farmer.chat:5438/", port=5438, grpc_port=5439, prefer_grpc=False)
    
    # Initialize OpenAI client
    api_key = os.getenv('OPENAI_API_KEY')
    openai_client = OpenAI(api_key=api_key)
    
    # Load the parsed QA chunks
    print("Loading parsed QA chunks...")
    # df = pd.read_csv("parsed_qa_chunks_with_retrieval.csv")
    df = pd.read_csv("/Users/ganeshnallagachu/Desktop/world_bank/chunk_data/comninational_english_final.csv")
    sample_df = df.head(1000)
    sample_df = sample_df[['chunk_id',"chunks","question","only_relevant_facts"]]

    df = sample_df
    # df = df.head(1000)
    print(f"Loaded {len(df)} questions")
    
    # Check if retrieval column already exists
    if 'retrieved_chunk_ids_k_5' in df.columns:
        print("Column 'retrieved_chunk_ids_k_5' already exists. Skipping retrieval.")
        return df
    
    # Get unique questions to avoid duplicate processing
    unique_questions = df['question'].unique()
    print(f"Found {len(unique_questions)} unique questions")
    
    # Collection name (based on your embedding_push copy.ipynb)
    collection_name = "test_agriculture_chunks"  # Change this to your actual collection name
    
    # Check if collection exists
    try:
        collections = qdrant_client.get_collections()
        collection_names = [col.name for col in collections.collections]
        if collection_name not in collection_names:
            print(f"Collection '{collection_name}' not found. Available collections: {collection_names}")
            return None
    except Exception as e:
        print(f"Error checking collections: {e}")
        return None
    
    # Parallel retrieval with optimized parameters
    print(f"Starting parallel retrieval with {len(unique_questions)} questions...")
    start_time = time.time()
    
    retrieval_results = parallel_retrieve_chunks(
        unique_questions, 
        qdrant_client, 
        collection_name, 
        openai_client, 
        max_workers=5,  # Adjust based on your system and API limits
        batch_size=50,   # Questions per batch
        top_k=5
    )
    
    end_time = time.time()
    print(f"Retrieval completed in {end_time - start_time:.2f} seconds")
    
    # Create mapping from question to retrieved chunks
    # question_to_chunks = dict(retrieval_results)
    
    # # Add retrieved chunks to the dataframe
    # df['retrieved_chunk_ids_k_5'] = df['question'].map(question_to_chunks)
    # df['retrieved_chunks_k_5'] = df['question'].map(question_to_chunks)

    # Create mapping from question to retrieved chunks
    question_to_chunk_ids = {}
    question_to_chunks = {}

    for question, chunk_ids, chunks_list in retrieval_results:
        question_to_chunk_ids[question] = chunk_ids
        question_to_chunks[question] = chunks_list

    # Add retrieved chunks to the dataframe
    df['retrieved_chunk_ids_k_5'] = df['question'].map(question_to_chunk_ids)
    df['retrieved_chunks_k_5'] = df['question'].map(question_to_chunks)
    
    # Save the updated dataframe
    output_file = "faithfullness_score.csv"
    df.to_csv(output_file, index=False)
    print(f"Saved updated dataframe to {output_file}")
    
    # # Display statistics
    # print("\nRetrieval Statistics:")
    # print(f"Total questions: {len(df)}")
    # print(f"Questions with retrieved chunks: {df['retrieved_chunk_ids'].notna().sum()}")
    # print(f"Questions without retrieved chunks: {df['retrieved_chunk_ids'].isna().sum()}")
    
    # # Show some examples
    # print("\nExample retrievals:")
    # for i, row in df.head().iterrows():
    #     print(f"Question: {row['question'][:50]}...")
    #     print(f"Original chunk IDs: {row['chunk_ids']}")
    #     print(f"Retrieved chunk IDs: {row['retrieved_chunk_ids']}")
        # print("-" * 50)
    
    return df


if __name__ == "__main__":
    # Run the main retrieval
    if __name__ == "__main__":
        main()
```

```python
import pandas as pd
import ast


def calculate_retrieval_accuracy(actual_chunks_str, retrieved_chunks_str):
    """
    Calculate retrieval accuracy as percentage of actual chunks found in retrieved chunks
    """
    try:
        # Parse string representations of lists to actual lists
        actual_chunks = ast.literal_eval(actual_chunks_str)
        retrieved_chunks = ast.literal_eval(retrieved_chunks_str)
        
        # Convert to sets for easier comparison
        actual_set = set(actual_chunks)
        retrieved_set = set(retrieved_chunks)
        
        # Find intersection (chunks that are both actual and retrieved)
        found_chunks = actual_set.intersection(retrieved_set)
        
        # Calculate accuracy percentage
        if len(actual_chunks) == 0:
            return 0.0  # Handle edge case of no actual chunks
        
        accuracy = (len(found_chunks) / len(actual_chunks)) * 100
        
        return accuracy
        
    except Exception as e:
        print(f"Error processing chunks: {e}")
        return 0.0

# Apply the function to calculate retrieval accuracy
results['retrieval_accuracy'] = results.apply(
    lambda row: calculate_retrieval_accuracy(row['chunk_ids'], row['retrieved_chunk_ids']), 
    axis=1
)

# Display results
print("Results with Retrieval Accuracy:")
print(results)
print(f"\nShape: {results.shape}")

# Show detailed breakdown for the first row
print("\nDetailed Analysis for first row:")
actual_chunks = ast.literal_eval(results['chunk_ids'].iloc[0])
retrieved_chunks = ast.literal_eval(results['retrieved_chunk_ids'].iloc[0])

print(f"Actual chunks: {actual_chunks}")
print(f"Retrieved chunks: {retrieved_chunks}")

found_chunks = set(actual_chunks).intersection(set(retrieved_chunks))
print(f"Found chunks: {list(found_chunks)}")
print(f"Missing chunks: {list(set(actual_chunks) - set(retrieved_chunks))}")
print(f"Accuracy: {results['retrieval_accuracy'].iloc[0]:.2f}%")

# Save to CSV
results.to_csv('results_with_accuracy.csv', index=False)
print(f"\nCSV file 'results_with_accuracy.csv' has been created!")
```

```python
import pandas as pd
import ast
import numpy as np
from typing import List, Set

def parse_chunk_ids(chunk_ids_str: str) -> Set[str]:
    """Parse chunk IDs from string representation to set."""
    try:
        # Remove quotes and parse as list
        chunk_ids_str = chunk_ids_str.strip()
        if chunk_ids_str.startswith('[') and chunk_ids_str.endswith(']'):
            # Parse as Python list
            chunk_list = ast.literal_eval(chunk_ids_str)
            return set(chunk_list)
        else:
            # Handle case where it might be comma-separated
            return set(chunk_ids_str.split(','))
    except:
        # If parsing fails, return empty set
        return set()

def calculate_precision_recall(true_chunks: Set[str], retrieved_chunks: Set[str]) -> tuple:
    """Calculate precision and recall for a single query."""
    if not retrieved_chunks:
        return 0.0, 0.0
    
    if not true_chunks:
        return 0.0, 0.0
    
    # Calculate intersection
    relevant_retrieved = true_chunks.intersection(retrieved_chunks)
    
    # Precision = relevant retrieved / total retrieved
    precision = len(relevant_retrieved) / len(retrieved_chunks)
    
    # Recall = relevant retrieved / total relevant
    recall = len(relevant_retrieved) / len(true_chunks)
    
    return precision, recall

def calculate_accuracy(true_chunks: Set[str], retrieved_chunks: Set[str]) -> float:
    """Calculate accuracy as exact match of retrieved chunks with true chunks."""
    if not true_chunks and not retrieved_chunks:
        return 1.0  # Both empty, perfect match
    
    if not true_chunks or not retrieved_chunks:
        return 0.0  # One empty, other not
    
    # Check if retrieved chunks exactly match true chunks
    return 1.0 if true_chunks == retrieved_chunks else 0.0

# Read the CSV file
df = pd.read_csv("parsed_qa_chunks_with_retrieval.csv")

print(f"Total queries: {len(df)}")
print("\n" + "="*50)

# Initialize metrics
total_precision = 0.0
total_recall = 0.0
total_accuracy = 0.0
valid_queries = 0

# Process each query
for idx, row in df.iterrows():
    question = row['question']
    true_chunks = parse_chunk_ids(row['chunk_ids'])
    retrieved_chunks = parse_chunk_ids(row['retrieved_chunk_ids'])
    
    # Skip if parsing failed
    if true_chunks is None or retrieved_chunks is None:
        continue
    
    # Calculate metrics
    precision, recall = calculate_precision_recall(true_chunks, retrieved_chunks)
    accuracy = calculate_accuracy(true_chunks, retrieved_chunks)
    
    # Accumulate metrics
    total_precision += precision
    total_recall += recall
    total_accuracy += accuracy
    valid_queries += 1
    
    # Print individual query results (optional)
    if idx < 5:  # Show first 5 queries as examples
        print(f"\nQuery {idx + 1}: {question[:100]}...")
        print(f"True chunks: {true_chunks}")
        print(f"Retrieved chunks: {retrieved_chunks}")
        print(f"Precision: {precision:.3f}, Recall: {recall:.3f}, Accuracy: {accuracy:.3f}")

# Calculate averages
if valid_queries > 0:
    avg_precision = total_precision / valid_queries
    avg_recall = total_recall / valid_queries
    avg_accuracy = total_accuracy / valid_queries
    
    print("\n" + "="*50)
    print("OVERALL METRICS:")
    print(f"Valid queries processed: {valid_queries}")
    print(f"Average Precision: {avg_precision:.4f}")
    print(f"Average Recall: {avg_recall:.4f}")
    print(f"Average Accuracy: {avg_accuracy:.4f}")
    
    # Calculate F1 Score
    if avg_precision + avg_recall > 0:
        f1_score = 2 * (avg_precision * avg_recall) / (avg_precision + avg_recall)
        print(f"F1 Score: {f1_score:.4f}")
    
    # Calculate percentage of perfect matches
    perfect_matches = sum(1 for i, row in df.iterrows() 
                        if calculate_accuracy(parse_chunk_ids(row['chunk_ids']), 
                                           parse_chunk_ids(row['retrieved_chunk_ids'])) == 1.0)
    perfect_match_rate = perfect_matches / valid_queries
    print(f"Perfect matches: {perfect_matches}/{valid_queries} ({perfect_match_rate:.2%})")
    
else:
    print("No valid queries found!")
```
